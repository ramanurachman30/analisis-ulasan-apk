# Pelabelan Sentimen Data Ulasan Gojek

## 1. Membaca Data Hasil Cleaning

```python
import pandas as pd

# 1. Membaca data hasil cleaning
file_clean = "data/data_step2_clean.csv"
df_final = pd.read_csv(file_clean)

jumlah_data_final = len(df_final)
print(f"1. Jumlah data bersih yang dibaca: {jumlah_data_final} baris")

# Sesuaikan nama kolom dengan dataset Anda jika berbeda
kolom_rating = 'score' # atau bisa jadi 'rating'

if kolom_rating not in df_final.columns:
    print(f"Perhatian: Kolom {kolom_rating} tidak ditemukan! Harap sesuaikan nama kolom.")
```

## 2. Distribusi Sentimen (Positif & Negatif)

```python
# 2. Distribusi Sentimen (Positif & Negatif)
# Positif: rating 4 & 5
# Negatif: rating 1 & 2
df_positif = df_final[df_final[kolom_rating] >= 4]
df_negatif = df_final[df_final[kolom_rating] <= 2]

jumlah_positif = len(df_positif)
jumlah_negatif = len(df_negatif)

persentase_positif = (jumlah_positif / jumlah_data_final) * 100 if jumlah_data_final > 0 else 0
persentase_negatif = (jumlah_negatif / jumlah_data_final) * 100 if jumlah_data_final > 0 else 0

print(f"\n2. Distribusi Sentimen pada Data Bersih:")
print(f"   - Positif (Rating 4-5): {jumlah_positif} baris ({persentase_positif:.2f}%)")
print(f"   - Negatif (Rating 1-2): {jumlah_negatif} baris ({persentase_negatif:.2f}%)")
```

## 3. Membuat Kolom Label dan Menyimpan Hasil Akhir

```python
# 3. Membuat kolom label untuk proses selanjutnya
# Peringatan (SettingWithCopyWarning) dapat dihindari dengan .copy()
df_final = df_final.copy()
df_final['sentiment'] = df_final[kolom_rating].apply(lambda x: 'positif' if x >= 4 else 'negatif')

# Menyimpan hasil akhir beserta labelnya
df_final.to_csv("data/data_step3_final_labeled.csv", index=False)
print("   -> Data bersih beserta kolom label disimpan sebagai: data_step3_final_labeled.csv")
```
