# Cleaning Data Ulasan Gojek

## 1. Menentukan Jumlah Data Awal

```python
import pandas as pd

# 1. Menentukan jumlah data awal
file_awal = "data/ulasan_gojek.csv"
df = pd.read_csv(file_awal)

jumlah_data_awal = len(df)
print(f"1. Jumlah data awal: {jumlah_data_awal} baris")

# Sesuaikan nama kolom dengan dataset Anda jika berbeda
kolom_rating = 'score' # atau bisa jadi 'rating'
kolom_teks = 'content' # kolom teks ulasan (untuk cek duplikat)

if kolom_rating not in df.columns:
    print(f"Perhatian: Kolom {kolom_rating} tidak ditemukan! Harap sesuaikan nama kolom.")
```

## 2. Menghapus Data dengan Rating 3 (Ambigu/Netral)

```python
# 2. Menghapus data dengan rating 3 (ambigu/netral)
df_no_rating3 = df[df[kolom_rating] != 3]
jumlah_dihapus_rating3 = jumlah_data_awal - len(df_no_rating3)

print(f"2. Jumlah data rating 3 yang dihapus: {jumlah_dihapus_rating3} baris")

# Simpan ke CSV untuk step ini
df_no_rating3.to_csv("data/data_step1_no_rating3.csv", index=False)
print("   -> Disimpan sebagai: data_step1_no_rating3.csv")
```

## 3. Menghapus Data Duplikat dan Baris Kosong (NaN)

```python
# 3. Menghapus data duplikat dan baris yang kosong/NaN pada kolom teks
# Hapus data yang teks ulasannya kosong (NaN)
df_clean = df_no_rating3.dropna(subset=[kolom_teks])
jumlah_dihapus_nan = len(df_no_rating3) - len(df_clean)

# Hapus duplikat teks sama & label sama (simpan 1 data)
df_dedup_same_label = df_clean.drop_duplicates(subset=[kolom_teks, kolom_rating], keep='first')
jumlah_dihapus_duplikat_sama = len(df_clean) - len(df_dedup_same_label)

# Hapus duplikat teks sama & label berbeda (hapus seluruhnya)
df_final = df_dedup_same_label.drop_duplicates(subset=[kolom_teks], keep=False)
jumlah_dihapus_duplikat_beda = len(df_dedup_same_label) - len(df_final)

total_dihapus_tahap2 = jumlah_dihapus_nan + jumlah_dihapus_duplikat_sama + jumlah_dihapus_duplikat_beda
print(f"3. Total data tahap 2 yang dihapus: {total_dihapus_tahap2} baris")
print(f"   - Data teks kosong (NaN): {jumlah_dihapus_nan} baris")
print(f"   - Data duplikat (teks & label sama): {jumlah_dihapus_duplikat_sama} baris")
print(f"   - Data ambigu (teks sama & label beda dihapus semua): {jumlah_dihapus_duplikat_beda} baris")

# Simpan ke CSV untuk step ini
df_final.to_csv("data/data_step2_clean.csv", index=False)
print("   -> Disimpan sebagai: data_step2_clean.csv")
```

## 4. Dataset Final yang Bersih

```python
# 4. Dataset final yang benar-benar bersih
jumlah_data_final = len(df_final)
print(f"4. Jumlah data final yang bersih: {jumlah_data_final} baris")
```
